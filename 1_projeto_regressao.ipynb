### 2.4) Fazendo engenharia de variáveis 

```python
# Temos variáveis temporais no conjunto de dados 
# Podemos fazer as seguintes tranformações para melhorar o aprendizado do algoritmo 

# Idade da casa no momento da venda (Depreciação principal)
df['HouseAge'] = df['YrSold'] - df['YearBuilt']

# Anos desde a última reforma no momento da venda (Valorização por modernização)
df['RemodAge'] = df['YrSold'] - df['YearRemodAdd']

# Criando uma variável que indica se a casa já foi reformada 
df['IsRemodeled'] = (df['YearBuilt'] != df['YearRemodAdd']).astype(int)

# Criando uma variável que indica se a casa foi vendida no ano em que foi construída
df['IsNewHouse'] = (df['YrSold'] == df['YearBuilt']).astype(int) 

# Criando estações do ano (EUA: Primavera/Verão aquecidos)
def get_season(month):
    if month in [3, 4, 5]: return 'Spring'
    elif month in [6, 7, 8]: return 'Summer'
    elif month in [9, 10, 11]: return 'Autumn'
    else: return 'Winter'

df['SeasonSold'] = df['MoSold'].apply(get_season)

# Eliminando as variáveis utilizadas 

df = df.drop(columns=['YrSold', 'YearBuilt', 'YearRemodAdd', 'YearBuilt', 'MoSold' ])
```

### 2.6) Verificando valores extremos 

```python
# Verificando valores extremos - variáveis contínuas 

import pandas as pd

def verifica_valor_extremo(df, vars_list):
    resultados = []
    total_linhas = len(df)
    
    for var in vars_list:
        df_var = df[var].dropna() # Remove nulos para não distorcer os quartis
        Q1 = df_var.quantile(0.25)
        Q3 = df_var.quantile(0.75)
        IQR = Q3 - Q1
        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR
        
        # Conta a quantidade de outliers
        qtd_outliers = ((df_var < limite_inferior) | (df_var > limite_superior)).sum()
        
        # Calcula a porcentagem em relação ao total de linhas do dataset
        pct_outliers = (qtd_outliers / total_linhas) * 100
        
        resultados.append({
            'Variável': var,
            'Qtd_Outliers': qtd_outliers,
            'Percentual_Outliers (%)': round(pct_outliers, 2)
        })
    
    # Converte em DataFrame e ordena pelos maiores percentuais
    df_resultado = pd.DataFrame(resultados)
    return df_resultado.sort_values(by='Percentual_Outliers (%)', ascending=False).reset_index(drop=True)
```

```python
# Como não há porcentagem maior que 20%, iremos tratar os valores extremos imputando seu valor pela mediana. 

df_outliers = verifica_valor_extremo(df, df_cont.columns)
df_outliers
```

```python
# Vamos separar as variáveis de acordo com seu tipo e característica

# 1. Separar o Target e ID  
X = df.drop(columns=['Id', 'SalePrice'], errors='ignore')

# 2. Variáveis Categóricas Ordinais 
var_cat_ord = ['ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond', 'HeatingQC', 'KitchenQual', 'GarageQual', 'GarageCond', 'BsmtExposure', 'BsmtFinType1', 
                'BsmtFinType2', 'Functional', 'GarageFinish']
var_cat_ord = [col for col in var_cat_ord if col in X.columns] # Garantir que só pegamos as ordinais que realmente existem no dataset

# 3. Variáveis Categóricas Nominais
var_cat_nom = X.select_dtypes(include=['object', 'string']).columns.difference(var_cat_ord).tolist()

# 4. Numéricas (Contínuas e Discretas juntas para imputação/scaling)
LIMITE = 15 
num = X.select_dtypes(include=['int64', 'float64']).nunique()
var_num = num[num>LIMITE].index

# 5. Demais variáveis por serem categóricas representadas com números inteiros, serão usadas diretamente no modelo. 
```

### 3.1) Pipeline de transformação

```python
import pandas as pd 
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PowerTransformer, OneHotEncoder
from class_outlier import OutlierReplacerIQR

pipeline_cont = make_pipeline(SimpleImputer(strategy='median'), OutlierReplacerIQR(), PowerTransformer(method='yeo-johnson', standardize=True))
pipeline_cat_nom = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(drop='if_binary', handle_unknown='ignore', sparse_output=False))

ct = make_column_transformer((pipeline_cont, var_num), (pipeline_cat_nom, var_cat_nom), remainder='passthrough')
preprocessador = make_pipeline(ct, verbose=False)
```

```python
# Conferindo as transformações
preprocessador.fit(X)
```

```python
feature_names = preprocessador.get_feature_names_out()
transformed_data_array = preprocessador.transform(X)
transformed_df = pd.DataFrame(transformed_data_array, columns=feature_names)

print(transformed_df.dtypes.value_counts())

transformed_df.head()
```

## 4) Modelagem 

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PowerTransformer

df_variaveis = df.drop(columns=['Id', 'SalePrice'])

x_treino, x_teste, y_treino, y_teste = train_test_split(df_variaveis, df['SalePrice'], test_size=0.20, random_state=5)

target_norm = PowerTransformer(standardize=True)
# Ajusta no treino
y_treino_norm = target_norm.fit_transform(y_treino.values.reshape(-1, 1))
# Apenas transforma o teste (usando o aprendizado do treino)
y_teste_norm = target_norm.transform(y_teste.values.reshape(-1, 1))
```

### 4.1) Regressão linear múltipla - modelo base 

```python
from sklearn.linear_model import LinearRegression

modelo_base = LinearRegression()
pipeline_final = make_pipeline(preprocessador, modelo_base)

# Etapa de treino 
pipeline_final.fit(x_treino, y_treino_norm) 
```

```python
# Fazendo as previsões e avaliando o modelo 

y_pred_norm = pipeline_final.predict(x_teste) 
y_pred_real = target_norm.inverse_transform(y_pred_norm)
```

```python
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

r2 = r2_score(y_teste, y_pred_real)
mae = mean_absolute_error(y_teste, y_pred_real)
rmse = np.sqrt(mean_squared_error(y_teste, y_pred_real))

print("R² = %.2f %%" % (r2 * 100))
print("MAE = USD %.2f" % mae)
print("RMSE = USD %.2f" % rmse)
```

- O modelo consegue explicar **86,97%** da variação nos preços das casas usando as variáveis do dataset.

- Se uma casa vale US$ 200.000, um erro de US$ 16.000 (MAE) representa uma margem de cerca de **8%**

### 4.3) Testes de validação de uma regressão 

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.stattools import durbin_watson

def avaliar_pressupostos_regressao(
    pipeline, 
    preprocessador, 
    target_transformer, 
    x_teste, 
    y_teste, 
    variaveis_selecionadas=None
):
    """
    Avalia os pressupostos estatísticos de um modelo de regressão linear.
    
    Parâmetros:
    -----------
    pipeline : sklearn.pipeline.Pipeline
        Pipeline do Scikit-Learn treinado.
    preprocessador : ColumnTransformer ou similar
        Objeto responsável pela transformação das features de entrada.
    target_transformer : PowerTransformer ou similar
        Objeto de transformação da variável resposta (com inverse_transform).
    x_teste : pandas.DataFrame ou numpy.ndarray
        Dados de entrada de teste.
    y_teste : pandas.Series ou numpy.ndarray
        Valores reais da variável resposta (escala original).
    variaveis_selecionadas : list, opcional
        Lista com nomes das colunas mantidas pelo modelo (ex: Lasso). 
        Se None, utiliza todas as colunas geradas pelo preprocessador.
        
    Retorna:
    --------
    dict: Dicionário contendo os resultados dos testes e o DataFrame do VIF.
    """
    
    # 1. Transformação dos dados e geração da matriz tratada
    x_teste_transf = preprocessador.transform(x_teste)
    nomes_colunas = preprocessador.get_feature_names_out()
    x_matriz_proc = pd.DataFrame(x_teste_transf, columns=nomes_colunas)

    # 2. Previsões e Cálculo dos Resíduos na escala real
    y_pred_norm = pipeline.predict(x_teste)
    if y_pred_norm.ndim == 1:
        y_pred_norm = y_pred_norm.reshape(-1, 1)
        
    y_pred_real = target_transformer.inverse_transform(y_pred_norm).ravel()
    
    # Garantindo formato de array unidimensional para y_teste
    y_true = y_teste.values if hasattr(y_teste, 'values') else y_teste
    residuos = y_true.ravel() - y_pred_real

    # 3. Filtragem das variáveis (Lasso/Seleção) para os testes de matriz
    if variaveis_selecionadas is not None:
        x_matriz_teste = x_matriz_proc[variaveis_selecionadas]
    else:
        x_matriz_teste = x_matriz_proc

    # --- TESTE 1: NORMALIDADE DOS RESÍDUOS (Shapiro-Wilk) ---
    print("="*50)
    print("1. TESTE DE NORMALIDADE (Shapiro-Wilk)")
    print("="*50)
    
    stat_shapiro, p_shapiro = stats.shapiro(residuos)
    print(f"Estatística de Teste: {stat_shapiro:.4f}")
    print(f"Valor-p: {p_shapiro:.4e}")
    if p_shapiro > 0.05:
        print("RESULTADO: Resíduos possuem distribuição NORMAL (p > 0.05).")
    else:
        print("RESULTADO: Resíduos NÃO seguem distribuição normal (p <= 0.05).")

    # --- TESTE 2: HOMOCEDASTICIDADE (Breusch-Pagan) ---
    print("\n" + "="*50)
    print("2. TESTE DE HOMOCEDASTICIDADE (Breusch-Pagan)")
    print("="*50)
    
    x_com_const = np.column_stack([np.ones(len(x_matriz_teste)), x_matriz_teste.values])
    lm_stat, p_bp, f_stat, f_p = het_breuschpagan(residuos, x_com_const)
    print(f"Estatística de Teste: {lm_stat:.4f}")
    print(f"Valor-p: {p_bp:.4e}")
    if p_bp > 0.05:
        print("RESULTADO: Homocedasticidade mantida (variância dos erros é constante).")
    else:
        print("RESULTADO: Heterocedasticidade detectada (variância dos erros oscila).")

    # --- TESTE 3: AUTOCORRELAÇÃO DOS ERROS (Durbin-Watson) ---
    print("\n" + "="*50)
    print("3. INDEPENDÊNCIA DOS RESÍDUOS (Durbin-Watson)")
    print("="*50)
    
    dw = durbin_watson(residuos)
    print(f"Estatística Durbin-Watson: {dw:.2f}")
    if 1.5 <= dw <= 2.5:
        print("RESULTADO: Os resíduos NÃO possuem autocorrelação significativa (Pressuposto mantido).")
    elif dw < 1.5:
        print("RESULTADO: Autocorrelação POSITIVA detectada nos resíduos.")
    else:
        print("RESULTADO: Autocorrelação NEGATIVA detectada nos resíduos.")

    # --- TESTE 4: MULTICOLINEARIDADE (VIF) ---
    print("\n" + "="*50)
    print("4. MULTICOLINEARIDADE (VIF - Variance Inflation Factor)")
    print("="*50)
    
    colunas_validas = x_matriz_teste.loc[:, x_matriz_teste.std() > 0]
    
    vif_list = []
    for i in range(colunas_validas.shape[1]):
        with np.errstate(divide='ignore', invalid='ignore'):
            vif_val = variance_inflation_factor(colunas_validas.values, i)
        vif_list.append(vif_val)

    vif_data = pd.DataFrame({"Variável": colunas_validas.columns, "VIF": vif_list})
    vif_sorted = vif_data.sort_values(by="VIF", ascending=False).reset_index(drop=True)
    vif_sorted["Status"] = vif_sorted["VIF"].apply(
        lambda x: "Multicolinearidade Alta" if np.isinf(x) or x > 10 else "OK"
    )

    print(vif_sorted.head(10).to_string(index=False))

    # Dicionário de retorno estruturado
    resultados = {
        'shapiro_stat': stat_shapiro,
        'shapiro_pvalue': p_shapiro,
        'breusch_pagan_stat': lm_stat,
        'breusch_pagan_pvalue': p_bp,
        'durbin_watson': dw,
        'vif_dataframe': vif_sorted,
        'residuos': residuos,
        'y_pred_real': y_pred_real
    }
    
    return resultados
```

* O modelo atinge um **$R^2$ de ~86%**, o que garante alta confiança ao determinar o *LTV (Loan-to-Value)* — a proporção do empréstimo em relação ao valor do imóvel.
* Saber que 86% da variação de preço é explicada objetivamente por esses 40 fatores protege a carteira de crédito contra avaliações superfaturadas ou fraudulentas.

#### D. Otimização de Reformas e Investimentos (*Fix & Flip*)
Como o modelo foca nas 40 variáveis mais relevantes, investidores conseguem simular o **Retorno sobre o Investimento (ROI)** de uma reforma antes de gastar dinheiro.

> * **Exemplo:** O modelo permite simular quanto o valor final da casa aumenta ao reformar o porão (`TotalBsmtSF`) versus adicionar uma vaga de garagem (`GarageCars`). O investidor foca o orçamento estritamente nas variáveis que o algoritmo provou que adicionam valor real ao imóvel.

### Conclusão Final

O modelo com **40 variáveis** é claramente a melhor escolha para implementação final/produção. Ele entrega praticamente a **mesma precisão** que o modelo com 211 colunas, porém com **muito mais estabilidade, menor tempo de execução e manutenção simplificada**.
